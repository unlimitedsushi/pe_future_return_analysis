```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
names_df = pd.read_csv('./sp500_companies.csv')
index = list(names_df['Symbol'].str.replace('.', '-').replace('FI', 'FISV'))
index.remove('WBA')
index.remove('IPG')
index.remove('K')
index.remove('PSKY')
```

## New Attempt with Trailing P/E

```python
# retrieving all prices
dates = ['2025-12-30','2025-09-30', '2025-06-30', '2025-03-31', '2024-12-30']

price_data = {}
for x in index:
    ticker = yf.Ticker(x)
    prices = {}
    for d in dates:
        try:
            price = ticker.history(start=d,end=pd.to_datetime(d) + pd.Timedelta(days=1), interval="1d")['Close'].values[0]
            prices[d] = price
        except Exception as e:
            print(f"Skipping {x}: {e}")
            continue
    price_data[x] = prices
```

```python
price_df = pd.DataFrame(price_data)
price_df
```

```python
earnings_data = {}
yearly_earnings_data = {}
for x in index:
    ticker = yf.Ticker(x)
    try:
        earnings = dict(ticker.quarterly_income_stmt.loc['Basic EPS'])
        yearly = dict(ticker.income_stmt.loc['Basic EPS'])
        earnings_data[x] = earnings
        yearly_earnings_data[x] = yearly

    except Exception as e:
        print(f"Skipping {x}: {e}")
        continue
    
```

```python
earnings_df = pd.DataFrame(earnings_data)
earnings_df.index = earnings_df.index.snap('QE')
earnings_df = earnings_df[earnings_df.index.year.isin([2024, 2025])].groupby(level=0).first().loc['2023-12-31':'2025-09-30']

preprocessed_earnings_df = earnings_df.loc['2024-12-31':'2025-09-30'].sort_index(ascending=False)
# .loc['2024-12-31':'2025-09-30'].sort_index(ascending=False)
```

```python
yearly_earnings_df = pd.DataFrame(yearly_earnings_data)
yearly_earnings_df.index = yearly_earnings_df.index.snap('QE')
yearly_earnings_df = yearly_earnings_df[yearly_earnings_df.index.year.isin([2024,2025])].groupby(level=0).first().sort_index(ascending=False)
yearly_earnings_df
```

```python
check_for_na = ['2025-09-30','2025-06-30', '2025-03-31', '2024-12-31']


for x in index:
    for i in check_for_na:
        if np.isnan(earnings_df[x].loc[i]):
            print(x, i)

```

```python
check_for_na = ['2025-09-30','2025-06-30', '2025-03-31', '2024-12-31']

for x in index:
    for i in check_for_na:
        if (np.isnan(earnings_df.loc[i, x])) and (i == '2025-09-30'):
            q_sum = sum(earnings_df[x].loc[check_for_na[1:]]) #sum of quarterly
            y_eps = yearly_earnings_df[x].loc['2025-09-30'] # yearly
            q_eps = y_eps - q_sum #quarterly eps
            earnings_df[x].loc[i] = q_eps
            # print(earnings_df.loc[i, x]) # to check for na
        elif (np.isnan(earnings_df.loc[i, x])) and (i == '2025-06-30'):
            q_sum = sum(earnings_df[x].loc[['2025-03-31', '2024-12-31', '2024-09-30']]) #sum of quarterly
            y_eps = yearly_earnings_df[x].loc['2025-06-30'] # yearly
            q_eps = y_eps - q_sum #quarterly eps
            earnings_df[x].loc[i] = q_eps
        

```

```python
quarters = ['2025-09-30','2025-06-30', '2025-03-31', '2024-12-31']
price_dates = ['2025-12-30','2025-09-30', '2025-06-30', '2025-03-31', '2024-12-30']

pe_ratio_data = {}
for x in index:
    dic = {}
    for q in quarters:
        if (np.isnan(earnings_df.loc[q, x])) and (q == quarters[-1]):
            eps_2024 = yearly_earnings_df.loc[q, x]
            price = price_df.loc[price_dates[-1], x]
            pe_ratio = price / eps_2024
        else:
            eps = earnings_df.loc[q, x]
            if q == quarters[-1]:
                price = price_df.loc[price_dates[-1], x]
            else:
                price = price_df.loc[q, x]
            if eps > 0:
                pe_ratio = price / (eps * 4)
            else:
                pe_ratio = 0
        dic[q] = pe_ratio
    pe_ratio_data[x] = dic
```

```python
pe_df = pd.DataFrame(pe_ratio_data)
```
